# 100-run uniqueness analysis for an arbitrary N-site system

This optional notebook performs the independent 100-run basis-optimization analysis described in the manuscript. 
Each run starts from the site basis but uses a different random order of the Givens rotations.

The analysis records the converged $V_{\rm ave}$ and all ordered diagonal energies $\epsilon_\mu$. 
Distinct ordered-energy sets at comparable $V_{\rm ave}$ indicate nonunique converged solutions in the practical screening sense used in the paper.

The default example below is the three-site intermediate-coupling condition of Fig. 4. 
The same notebook can be used for any $N\times N$ Hamiltonian by replacing `H`, `lambda_i`, and `cfg`. 

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from QME import Config, uniqueness_scan_system
from manuscript_models import three_site_model

OUT = Path("results")
OUT.mkdir(exist_ok=True)

# Default: Fig. 4 intermediate-coupling three-site model.
H = three_site_model(100.0, 100.0, 20.0, 20.0)
lambda_i = np.full(H.shape[0], 100.0)
cfg = Config(temperature_k=77.0, tau_g_ps=0.05, nmax=10_000, dt_ps=1.0e-4)
```

```python
columns, scan, bases = uniqueness_scan_system(
    H, lambda_i, cfg, n_runs=100, seed_start=0
)
np.savetxt(OUT / "Nsite_100run_scan.dat", scan, header=" ".join(columns))
print(columns)
print(scan)
```

```python
fig, ax = plt.subplots()
ax.scatter(scan[:, 1], scan[:, 3])
ax.set(xlabel="Random seed", ylabel=r"$V_{\rm ave}$ (cm$^{-1}$)")
fig.tight_layout()
fig.savefig(OUT / "Nsite_100run_Vave.pdf", bbox_inches="tight")
plt.show()

print("Rounded Vave distribution:")
values, counts = np.unique(np.round(scan[:, 3], 6), return_counts=True)
for v, c in zip(values, counts):
    print(v, c)

print("\nOrdered diagonal energies for each run are stored in columns:")
print(columns[4:])
```

The ordered diagonal-energy columns are considered together with $V_{\rm ave}$. 
A scalar $V_{\rm ave}$ alone can hide distinct stationary bases. 
For Fig. 4, the two stationary families have nearly identical $V_{\rm ave}$ but different ordered diagonal energies.
